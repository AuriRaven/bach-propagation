# Fundamentos Matemáticos de las Razones de Frecuencia en la Afinación Musical

En este cuaderno se examina, desde una perspectiva matemática y física, el fenómeno de
interferencia de ondas sonoras que subyace a la percepción de consonancia y disonancia en
los intervalos musicales. Este análisis constituye la base teórica sobre la cual se
fundamenta la codificación de relaciones armónicas empleada en el modelo de generación
secuencial desarrollado en la presente tesis ("Bach Propagation").

## Superposición de ondas armónicas

Toda nota musical puede modelarse, en primera aproximación, como una onda senoidal pura:

$$ y(t) = A \sin(2\pi f t + \varphi) $$

donde $A$ es la amplitud, $f$ la frecuencia fundamental expresada en Hercios (Hz) y
$\varphi$ la fase inicial. Cuando dos ondas de frecuencias $f_1$ y $f_2$ se propagan
simultáneamente en un mismo medio, el principio de superposición establece que la
perturbación resultante es la suma algebraica de ambas:

$$ y(t) = \sin(2\pi f_1 t) + \sin(2\pi f_2 t) $$

El carácter consonante o disonante de un intervalo depende, en gran medida, de la razón
$f_2 / f_1$. Cuando dicha razón puede expresarse como el cociente de dos enteros
pequeños, la onda resultante es periódica con un período relativamente corto, lo cual el
sistema auditivo tiende a interpretar como estabilidad. En cambio, cuando la razón se
aproxima a la unidad sin coincidir exactamente con ella, se produce el fenómeno de
**batimento** (*beating*), perceptible como una pulsación cíclica de la amplitud.

```python
import numpy as np
import plotly.graph_objects as go

# Plantilla visual y paleta de colores del cuaderno
TEMPLATE = "plotly_dark"
COLOR_CYAN = "#00FFFF"      # Primera onda componente
COLOR_MAGENTA = "#FF00FF"   # Segunda onda componente
COLOR_YELLOW = "#FFFF00"    # Reservado para la comparación de temperamentos

# Frecuencia de referencia: A3 = 220 Hz
F0 = 220.0
```

## La Quinta Justa (razón 3:2)

La quinta justa se define mediante la razón racional $f_2/f_1 = 3/2$. Al tratarse del
cociente de dos enteros pequeños, la onda compuesta $y(t) = \sin(2\pi f_1 t) + \sin(2\pi
f_2 t)$ es **exactamente periódica**, con período:

$$ T = \frac{2}{f_1} = \frac{3}{f_2} $$

Es decir, cada dos ciclos de $f_1$ coinciden exactamente con tres ciclos de $f_2$, y el
patrón de interferencia se repite de manera idéntica indefinidamente, sin modulación de
amplitud. Esta periodicidad corta y estable es la razón física por la cual el oído
humano percibe la quinta justa como el intervalo consonante por excelencia (después de la
octava).

## La Segunda Menor (razón 16:15)

La segunda menor, en cambio, corresponde a una razón $f_2/f_1 = 16/15$, muy próxima a la
unidad. Al ser $f_1$ y $f_2$ casi iguales, la superposición de ambas ondas no genera un
patrón simple, sino un fenómeno de batimento. Para demostrarlo analíticamente, se aplica
la identidad trigonométrica de la suma de senos:

$$ \sin A + \sin B = 2\, \sin\!\left(\frac{A+B}{2}\right) \cos\!\left(\frac{A-B}{2}\right) $$

Sustituyendo $A = 2\pi f_1 t$ y $B = 2\pi f_2 t$, la suma de las dos ondas se reescribe
como:

$$ y(t) = \underbrace{2\cos\!\left(2\pi \frac{f_1 - f_2}{2}\, t\right)}_{\text{envolvente}}
\; \underbrace{\sin\!\left(2\pi \frac{f_1 + f_2}{2}\, t\right)}_{\text{portadora}} $$

Esta expresión revela la estructura del batimento: la onda resultante puede
interpretarse como una **portadora** que oscila a la frecuencia promedio $(f_1+f_2)/2$
—perceptualmente indistinguible de un tono único—, cuya amplitud es modulada por una
**envolvente** de baja frecuencia $\cos\!\left(2\pi \tfrac{f_1-f_2}{2} t\right)$. Puesto
que el sistema auditivo (y cualquier detector de amplitud) responde al valor absoluto de
la envolvente, la frecuencia de batimento efectivamente percibida es:

$$ f_{\text{batimento}} = |f_1 - f_2| $$

ya que $|\cos(\cdot)|$ duplica la frecuencia aparente de la envolvente al rectificar sus
semiciclos negativos. Para $f_1 = 220\ \text{Hz}$, esto produce un batimento de
aproximadamente $14.67\ \text{Hz}$, dentro del rango de aspereza (*roughness*) percibido
por el oído humano como disonancia.

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Parámetros de tiempo (1 segundo de simulación de alta resolución)
t = np.linspace(0, 0.1, 2000)
f_fundamental = 440.0 # A4

# 1. Consonancia: Quinta Justa Perfecta (Razón 3:2)
f_quinta = f_fundamental * (3/2)
onda_consonante = np.sin(2 * np.pi * f_fundamental * t) + np.sin(2 * np.pi * f_quinta * t)

# 2. Disonancia: Segunda Menor (Razón 16:15)
f_segunda = f_fundamental * (16/15)
onda_disonante = np.sin(2 * np.pi * f_fundamental * t) + np.sin(2 * np.pi * f_segunda * t)

# Crear subplots
fig = make_subplots(rows=2, cols=1, 
                    subplot_titles=("Consonancia (Quinta Justa 3:2)", 
                                    "Disonancia con Batimento (Segunda Menor 16:15)"))

# Trazo Consonante (Cyan)
fig.add_trace(go.Scatter(x=t, y=onda_consonante, 
                         mode='lines', 
                         line=dict(color='#00FFFF', width=2),
                         name="3:2 (Consonante)"), row=1, col=1)

# Trazo Disonante (Magenta)
fig.add_trace(go.Scatter(x=t, y=onda_disonante, 
                         mode='lines', 
                         line=dict(color='#FF00FF', width=2),
                         name="16:15 (Disonante)"), row=2, col=1)

# Estilizado de la interfaz
fig.update_layout(
    template="plotly_dark",
    title="Análisis Geométrico de Interferencia de Ondas",
    height=700,
    showlegend=False,
    paper_bgcolor='#111111',
    plot_bgcolor='#111111'
)

fig.update_xaxes(title_text="Tiempo (s)", showgrid=True, gridwidth=1, gridcolor='#333333')
fig.update_yaxes(title_text="Amplitud", showgrid=True, gridwidth=1, gridcolor='#333333')

fig.show()
```

```python
# --- Quinta Justa (razón 3:2) ---
f1 = F0
f2 = F0 * 3 / 2  # 330 Hz

t = np.linspace(0, 0.05, 4000)  # ventana de 50 ms

y1 = np.sin(2 * np.pi * f1 * t)
y2 = np.sin(2 * np.pi * f2 * t)
y_sum = y1 + y2

fig_fifth = go.Figure()
fig_fifth.add_trace(go.Scatter(x=t * 1000, y=y1, name=f"Onda 1 ({f1:.1f} Hz)",
                                line=dict(color=COLOR_CYAN, width=1)))
fig_fifth.add_trace(go.Scatter(x=t * 1000, y=y2, name=f"Onda 2 ({f2:.1f} Hz)",
                                line=dict(color=COLOR_MAGENTA, width=1)))
fig_fifth.add_trace(go.Scatter(x=t * 1000, y=y_sum, name="Interferencia (suma)",
                                line=dict(color="white", width=2)))

fig_fifth.update_layout(
    template=TEMPLATE,
    title="Interferencia de Ondas: Quinta Justa (Razón 3:2)",
    xaxis_title="Tiempo (ms)",
    yaxis_title="Amplitud",
)
fig_fifth.show()

# --- Segunda Menor (razón 16:15) ---
f1 = F0
f2 = F0 * 16 / 15  # ≈ 234.67 Hz

t = np.linspace(0, 0.2, 8000)  # 200 ms, suficientes para varios ciclos de batimento

y1 = np.sin(2 * np.pi * f1 * t)
y2 = np.sin(2 * np.pi * f2 * t)
y_sum = y1 + y2
envolvente = 2 * np.abs(np.cos(2 * np.pi * (f1 - f2) / 2 * t))

fig_second = go.Figure()
fig_second.add_trace(go.Scatter(x=t * 1000, y=y1, name=f"Onda 1 ({f1:.1f} Hz)",
                                 line=dict(color=COLOR_CYAN, width=1)))
fig_second.add_trace(go.Scatter(x=t * 1000, y=y2, name=f"Onda 2 ({f2:.2f} Hz)",
                                 line=dict(color=COLOR_MAGENTA, width=1)))
fig_second.add_trace(go.Scatter(x=t * 1000, y=y_sum, name="Interferencia (suma)",
                                 line=dict(color="white", width=1.5)))
fig_second.add_trace(go.Scatter(x=t * 1000, y=envolvente, name="Envolvente (batimento)",
                                 line=dict(color="white", width=1.5, dash="dash")))
fig_second.add_trace(go.Scatter(x=t * 1000, y=-envolvente, name="Envolvente (batimento)",
                                 line=dict(color="white", width=1.5, dash="dash"),
                                 showlegend=False))

fig_second.update_layout(
    template=TEMPLATE,
    title="Interferencia de Ondas: Segunda Menor (Razón 16:15)",
    xaxis_title="Tiempo (ms)",
    yaxis_title="Amplitud",
)
fig_second.show()
```

## Interpretación de los resultados

En la primera gráfica se observa que la onda de interferencia (trazo blanco) resultante
de la Quinta Justa reproduce un patrón que se repite idénticamente cada $T = 2/f_1$
segundos, sin variación de amplitud: es la firma visual de la consonancia.

En la segunda gráfica, en cambio, la envolvente (trazo discontinuo) delimita
explícitamente los máximos de amplitud de la onda suma, evidenciando el batimento
predicho por la identidad trigonométrica. La distancia temporal entre dos máximos
sucesivos de la envolvente corresponde a $1/f_{\text{batimento}} \approx 68.2\ \text{ms}$,
en concordancia con el valor teórico de $14.67\ \text{Hz}$ calculado anteriormente.

## Quinta Justa versus Quinta del Temperamento Igual

El análisis anterior asumió una afinación de **entonación justa** (*just intonation*),
en la que los intervalos se definen mediante razones racionales exactas. Sin embargo, la
práctica musical occidental moderna —incluida la obra de Johann Sebastian Bach y, en
particular, *El Clave Bien Temperado*— se apoya en el **temperamento igual** (12-TET),
en el cual la octava se divide en doce semitonos logarítmicamente equidistantes. Cada
semitono corresponde entonces a una razón de frecuencia:

$$ r_{\text{semitono}} = 2^{1/12} $$

y la quinta del temperamento igual, compuesta por siete semitonos, corresponde a:

$$ r_{ET} = 2^{7/12} \approx 1.498307077\ldots $$

un número **irracional**, en contraste directo con la quinta justa de razón exactamente
racional $r_{JI} = 3/2 = 1.5$. Esta irracionalidad no es una imprecisión de cálculo, sino
una consecuencia matemática ineludible. Si se supusiera, por reducción al absurdo, que
$2^{7/12} = 3/2$, entonces:

$$ 2^{7/12} = \frac{3}{2} \;\Longrightarrow\; 2^{7} = \left(\frac{3}{2}\right)^{12}
\;\Longrightarrow\; 2^{19} = 3^{12} $$

lo cual contradice el teorema fundamental de la aritmética, pues el miembro izquierdo es
una potencia exclusiva de $2$ y el miembro derecho una potencia exclusiva de $3$. Queda
así demostrado que, salvo la octava, **ningún intervalo del temperamento igual puede
coincidir exactamente con su contraparte de entonación justa**.

La diferencia entre $r_{JI}$ y $r_{ET}$, si bien minúscula (del orden de $-1.96$ *cents*),
produce un batimento cuya frecuencia $\Delta f = f_0 \left(r_{JI} - r_{ET}\right)$ es
extremadamente baja —del orden de fracciones de Hercio para frecuencias del registro
medio—. Por esta razón, dicho batimento resulta imperceptible en una ventana temporal
corta como las empleadas anteriormente, y solo se manifiesta al extender
suficientemente el vector de tiempo. La siguiente gráfica ilustra este **batimento
micrométrico**, evidencia sonora directa de la irracionalidad de la afinación moderna.

```python
# --- Comparación: Quinta Justa (3:2) vs Quinta del Temperamento Igual (2^(7/12)) ---
f_just = F0 * 3 / 2         # 330.000 Hz (entonación justa)
f_et = F0 * 2 ** (7 / 12)   # ≈ 329.628 Hz (temperamento igual)

delta_f = abs(f_just - f_et)
beat_period = 1 / delta_f

# Vector de tiempo extendido lo suficiente para revelar el batimento micrométrico
n_beats = 3
t_max = n_beats * beat_period
fs = 1500  # Hz de muestreo, suficiente para resolver ambas frecuencias de audio
t = np.linspace(0, t_max, int(fs * t_max))

y_just = np.sin(2 * np.pi * f_just * t)
y_et = np.sin(2 * np.pi * f_et * t)
envolvente = 2 * np.abs(np.cos(2 * np.pi * delta_f / 2 * t))

fig_temper = go.Figure()
fig_temper.add_trace(go.Scatter(x=t, y=y_just, name=f"Quinta Justa ({f_just:.3f} Hz)",
                                 line=dict(color=COLOR_CYAN, width=0.8)))
fig_temper.add_trace(go.Scatter(x=t, y=y_et, name=f"Quinta Temperada ({f_et:.3f} Hz)",
                                 line=dict(color=COLOR_MAGENTA, width=0.8)))
fig_temper.add_trace(go.Scatter(x=t, y=envolvente, name="Envolvente de batimento",
                                 line=dict(color=COLOR_YELLOW, width=2, dash="dash")))
fig_temper.add_trace(go.Scatter(x=t, y=-envolvente, name="Envolvente de batimento",
                                 line=dict(color=COLOR_YELLOW, width=2, dash="dash"),
                                 showlegend=False))

fig_temper.update_layout(
    template=TEMPLATE,
    title=(f"Quinta Justa vs. Quinta Temperada — Batimento de {delta_f:.4f} Hz "
           f"(período ≈ {beat_period:.2f} s)"),
    xaxis_title="Tiempo (s)",
    yaxis_title="Amplitud",
)
fig_temper.show()

cents = 1200 * np.log2(f_et / f_just)
print(f"Razón justa (3/2):            {3/2:.10f}")
print(f"Razón temperada (2^(7/12)):   {2 ** (7 / 12):.10f}")
print(f"Diferencia:                   {cents:.4f} cents")
print(f"Frecuencia de batimento:      {delta_f:.4f} Hz")
print(f"Período de batimento:         {beat_period:.4f} s")
```

## Conclusión

Los tres casos analizados —Quinta Justa, Segunda Menor y la comparación entre Quinta
Justa y Quinta Temperada— ilustran un mismo principio físico bajo tres manifestaciones
distintas de la razón $f_2/f_1$: periodicidad exacta cuando la razón es racional y
simple, batimento rápido y perceptible cuando la razón racional es cercana a la unidad, y
batimento lento e imperceptible al oído inmediato —pero matemáticamente inevitable—
cuando la razón es irracional, como ocurre en el temperamento igual. Esta última
observación resulta especialmente relevante para "Bach Propagation": toda codificación
simbólica de altura basada en clases de tono (*pitch classes*) del temperamento igual
descarta, por construcción, esta microestructura de afinación, una simplificación que
conviene tener presente al interpretar los resultados del modelo frente a la práctica
interpretativa histórica.
